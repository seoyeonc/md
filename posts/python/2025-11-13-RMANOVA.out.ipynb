# Import

```python
import pandas as pd
import numpy as np

import pingouin as pg

import matplotlib.pyplot as plt
import seaborn as sns
```

# Data

```python
np.random.seed(1212)
```

```python
n_groups = 3
n_sub_per_group = 50
groups = ['G1', 'G2', 'G3']
times = ['T1', 'T2', 'T3'] 
```

```python
data = []
subject_id = 1
for g in groups:
    for _ in range(n_sub_per_group):
        subject_effect = np.random.normal(0, 3)
        
        if g == 'G1':
            base_mean = 50
        elif g == 'G2':
            base_mean = 55
        else:  # G3
            base_mean = 60
            
        means_by_time = {
            'T1': base_mean,
            'T2': base_mean + 5,
            'T3': base_mean + 8
        }
        
        for t in times:
            score = np.random.normal(means_by_time[t] + subject_effect, 5)
            data.append([subject_id, g, t, score])
        
        subject_id += 1
```

```python
df = pd.DataFrame(data, columns=['subject', 'group', 'time', 'score'])
```

```python
df.head()
```

```python
df.groupby(['group', 'time'])['score'].mean()
```

# Repeated Measure ANOVA

```python
anova = pg.mixed_anova(
    data=df,
    dv='score',
    within='time',
    between='group',
    subject='subject'
)
```

```python
anova
```

# Results

1.  Between-subject effect: group

-   F(2,147)=96.43, p \< .0001
-   The mean differences among the three groups were statistically
    significant.
-   np2 - 0.567 -\> Large effect
-   The mean baselines of three groups are different.
-   The Groups have effective to the Score.

1.  Within-subject effect: time (Repeated measure factor)

-   F(2,294)=78.37, p \< .0001
-   The mean scores changed significantly over time.
-   p = 5.24e-28 -\> significant
-   np2 = 0.348
-   eps = 1 -\> sphericity satisfation

1.  Interaction effect: group × time

-   F(4,294)=2.64, p = 0.034 -\> significant

-   p = 0.034 -\> significant

-   np2 = 0.035 -\> small effect

-   This indicates that the pattern of change over time differs across
    groups.

-   In the mixed ANOVA, the main effect of group was significant
    (F(2,147) = 96.43, p \< .001, η² = .57), indicating substantial
    differences among the three groups.

-   The main effect of time was also significant (F(2,294) = 78.37, p \<
    .001, η² = .35), showing that scores changed meaningfully across the
    three time points.

-   Additionally, the group × time interaction was significant (F(4,294)
    = 2.64, p = .034), suggesting that the pattern of change over time
    differed across groups.
