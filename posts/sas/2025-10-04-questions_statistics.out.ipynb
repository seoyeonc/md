# 1. Technical / Knowledge

## 1.1. Difference between ADaM and SDTM

| Aspect           | **SDTM**                                               | **ADaM**                                                                                  |
|--------|-------------------------|----------------------------------------|
| Full name        | Study Data Tabulation Model                            | Analysis Data Model                                                                       |
| Purpose          | Standardized datasets for **regulatory submission**    | Derived datasets for **statistical analysis & TFLs**                                      |
| Data type        | Close to **raw collected data**                        | **Processed/derived data** from SDTM                                                      |
| Structure        | Organized by **domains** (DM, AE, LB, VS, etc.)        | Organized by **analysis datasets** (ADSL, ADAE, ADLB, etc.)                               |
| Key features     | Ensures **traceability** and consistency for reviewers | Includes **derived variables**, **analysis flags**, and is ready for SAP-defined analyses |
| Primary users    | **Regulatory agencies** (FDA, EMA reviewers)           | **Biostatisticians & Statistical Programmers**                                            |
| One-line summary | “Submission-ready data”                                | “Analysis-ready data”                                                                     |

# 3. Statistical Thinking & Problem-Solving

## 3.1. a Type I and Tpe II error in clinical trials

|                                  | **True (Reality)**       | **False (Reality)**  |
|------------------------|------------------------|-------------------------|
| **Positive (Reject H₀)**         | Correct decision (Power) | **Type I Error (α)** |
| **Negative (Fail to Reject H₀)** | **Type II Error (β)**    | Correct decision     |

| Aspect                     | **Type I Error**                                                             | **Type II Error**                                                                |
|--------------|-----------------------------|------------------------------|
| **Also called**            | False Positive                                                               | False Negative                                                                   |
| **Symbol**                 | α (alpha)                                                                    | β (beta)                                                                         |
| **Definition**             | Concluding that a treatment has an effect when it actually does not          | Failing to detect a true treatment effect that actually exists                   |
| **Clinical trial example** | A new drug is declared effective when it is not → unnecessary approval       | A truly effective drug is rejected as ineffective → missed treatment opportunity |
| **Control method**         | Set significance level (e.g., α = 0.05)                                      | Ensure sufficient statistical power (1 – β, e.g., 80%)                           |
| **Clinical importance**    | Prevents false claims of efficacy; highly controlled by regulators (FDA/EMA) | Avoids missing truly beneficial therapies                                        |
| **Key phrase**             | Type I error = claiming a false effect                                       | Type II error = missing a true effect                                            |

## 3.3. Reason for Interim analysis in clinical trials

> 임상 결과 우연하게 좋게 결과가 나온다면 효과가 없는데 있다는 flase
> positive type I error가 나올 수 있다. 이를 방짐라기 위해 임상시험에서
> 결과를 판단하는 기준인 유의수준$\alpha$를 나눈 alpha spending 방법을
> 써서(예: 중간분석 0.01, 최종분석 0.04, 누적 0.05)로 알파를 0.05
> 초과하지 않게 유지해서 확인. 이때 알파를 나누는 방법으로는 O’Brien
> fleming, Pocock 이 있다.

-   Purpose
    -   assess safety, efficacy, or futility
    -   avoid overall Type I error inflation
-   Method to control overall Type I error
    -   use group sequential designs or alpha-spending methods
        -   devide the total significance level across interim and final
            analyses so that the overall false-positive risk stays
            within 5%.
    -   use O’Brien-Fleming, Pocock methods

## 3.4. Survival Analysis(Survival data or time-to-time data)

1.  **a Kaplan-Meier** analysis to estimate survival probabilities and
    visualize survival curves for each group.
    -   생존 확률 추정 + considering censoring
2.  **a Log-rank test** to compare whether ther is a significant
    difference between the treatment and control groups.
    -   군 간 생존률 비교
3.  **a Cox proportional hazards model** to estimate the hazard ratio
    and its confidence interval to quantify the treatment effect.
    -   위험비 추정 + adjust for covariates
