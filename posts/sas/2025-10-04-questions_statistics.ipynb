# Basic Abbreviation

| Term  | Abbreviation                                   | Definition                                                                                  |
| ----- | ---------------------------------------------- | ------------------------------------------------------------------------------------------- |
| CDISC | Clinical Data Interchange Standards Consortium | An international non-profit organization that develops data standards for clinical research |
| SDTM  | Study Data Tabulation Model                    | A standard for organizing and submitting clinical trial data                                |
| ADaM  | Analysis Data Model                            | A standard for creating analysis-ready datasets in clinical trials                          |

# 1. Technical / Knowledge

## 1.1. Difference between ADaM and SDTM

| Aspect           | **SDTM**                                               | **ADaM**                                                                                  |
| ---------------- | ------------------------------------------------------ | ----------------------------------------------------------------------------------------- |
| Full name        | Study Data Tabulation Model                            | Analysis Data Model                                                                       |
| Purpose          | Standardized datasets for **regulatory submission**    | Derived datasets for **statistical analysis & TFLs**                                      |
| Data type        | Close to **raw collected data**                        | **Processed/derived data** from SDTM                                                      |
| Structure        | Organized by **domains** (DM, AE, LB, VS, etc.)        | Organized by **analysis datasets** (ADSL, ADAE, ADLB, etc.)                               |
| Key features     | Ensures **traceability** and consistency for reviewers | Includes **derived variables**, **analysis flags**, and is ready for SAP-defined analyses |
| Primary users    | **Regulatory agencies** (FDA, EMA reviewers)           | **Biostatisticians & Statistical Programmers**                                            |
| One-line summary | “Submission-ready data”                                | “Analysis-ready data”                                                                     |

## 1.2. Derived dataset(ADaM), Derived Variables

Meta-analysis를 할 때 주의해야 할 점은 무엇인가요?
