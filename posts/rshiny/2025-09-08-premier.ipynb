## Premier League Top 6 Teams Match Statistics

- [link](https://seoyeonchoi.shinyapps.io/premier/)
- [data](https://www.kaggle.com/datasets/ensarakbas/premier-league-top-6-teams-match-statistics)

| 열 이름                   | 의미 |
|---------------------------|------|
| Date                      | 경기 날짜와 시간 |
| Opponent                  | 상대 팀 이름 |
| Is_Home                   | 홈경기 여부 (1=홈, 0=원정) |
| Result                    | 경기 결과 (1=승, 0=무, -1=패) |
| Goals                     | 우리 팀이 넣은 골 수 |
| Opponent_Goals            | 상대 팀이 넣은 골 수 |
| Possession                | 우리 팀 점유율 (%) |
| Shots                     | 슈팅 개수 |
| Shots_On_Target           | 유효 슈팅 개수 |
| Passes_Completed          | 성공한 패스 수 |
| Pass_Accuracy             | 패스 성공률 (%) |
| Corners                   | 코너킥 수 |
| Crosses                   | 크로스 수 |
| Fouls                     | 반칙 수 |
| Offsides                  | 오프사이드 수 |
| Opponent_Possession       | 상대 팀 점유율 (%) |
| Opponent_Shots            | 상대 슈팅 개수 |
| Opponent_Shots_On_Target  | 상대 유효 슈팅 개수 |
| Opponent_Passes_Completed | 상대 성공 패스 수 |
| Opponent_Pass_Accuracy    | 상대 패스 성공률 (%) |
| Opponent_Corners          | 상대 코너킥 수 |
| Opponent_Crosses          | 상대 크로스 수 |
| Opponent_Fouls            | 상대 반칙 수 |
| Opponent_Offsides         | 상대 오프사이드 수 |
| Shot_Efficiency           | 슈팅 효율 (Goals / Shots) |
| Season                    | 시즌 연도 |
| Month                     | 경기 월 |
| Day_of_Week               | 경기 요일 (1=월 … 7=일) |
| Last5_Avg_Goals           | 최근 5경기 평균 득점 |
| Last5_Win_Rate            | 최근 5경기 승률 |
| team                      | 데이터 소속 팀 |
